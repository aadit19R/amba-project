```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt

df = pd.read_csv("Boston.csv")
df["Crim_log"] = np.log1p(df["crim"])
display(df[["crim", "Crim_log"]].head())

sns.histplot(df["Crim_log"], kde=True)
plt.show()
```

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Run the data-loading cell first: df = pd.read_csv("Boston.csv")
# Natural log is defined only for finite, positive values.
crim = pd.to_numeric(df['crim'], errors='coerce')
valid = np.isfinite(crim) & (crim > 0)
df['crim_log'] = np.log(crim.where(valid))
crim_log = df['crim_log'].dropna()

if crim_log.empty:
    raise ValueError("The crim column has no finite, positive values to log-transform.")

# Calculate Q1, Q3, and IQR on valid log-transformed values.
Q1 = crim_log.quantile(0.25)
Q3 = crim_log.quantile(0.75)
IQR = Q3 - Q1

lower_limit = Q1 - 1.5 * IQR
upper_limit = Q3 + 1.5 * IQR

outliers = df[
    (df['crim_log'] < lower_limit) |
    (df['crim_log'] > upper_limit)
]

print("Column: crim_log")
print("Number of outliers:", len(outliers))
print(f"Lower limit: {lower_limit:.2f}")
print(f"Upper limit: {upper_limit:.2f}")
print("Values excluded from log transformation:", int((~valid).sum()))
print("\nOutliers:")
print(outliers)

plt.figure(figsize=(7, 3))
sns.boxplot(x=crim_log)
plt.title("Box Plot - crim_log")
plt.xlabel("crim_log")
plt.tight_layout()
plt.show()
```

```python
import numpy as np

for col in ['dis', 'black', 'lstat']:
    m = np.mean(df[col])
    sd = np.std(df[col])
    lower, upper = m - 3 * sd, m + 3 * sd
    print(f'{col}: 3σ bounds = [{lower:.2f}, {upper:.2f}], outliers = {((df[col] < lower) | (df[col] > upper)).sum()}')
```

```python
import numpy as np

m = np.mean(df['dis'])
sd = np.std(df['dis'])
ul_m3s = m + 3 * sd
df['dis1'] = np.where(df['dis'] > ul_m3s, ul_m3s, df['dis'])

print(df.loc[df['dis'] > ul_m3s, ['dis', 'dis1']])
```

```python
import numpy as np

m = np.mean(df['black'])
sd = np.std(df['black'])
ul_m3s = m + 3 * sd
df['black1'] = np.where(df['black'] > ul_m3s, ul_m3s, df['black'])

changes = df.loc[df['black'] > ul_m3s, ['black', 'black1']]
print(changes if not changes.empty else "No black values exceed mean + 3σ; no changes.")
```

```python
import numpy as np

m = np.mean(df['lstat'])
sd = np.std(df['lstat'])
ul_m3s = m + 3 * sd
df['lstat1'] = np.where(df['lstat'] > ul_m3s, ul_m3s, df['lstat'])

changes = df.loc[df['lstat'] > ul_m3s, ['lstat', 'lstat1']]
print(changes if not changes.empty else "No lstat values exceed mean + 3σ; no changes.")
```

```python
import numpy as np

Q1, Q3 = df['lstat'].quantile([0.25, 0.75])
IQR = Q3 - Q1
lower, upper = Q1 - 1.5 * IQR, Q3 + 1.5 * IQR
df['lstat_iqr'] = np.where(df['lstat'] < lower, lower,
                           np.where(df['lstat'] > upper, upper, df['lstat']))
print(df.loc[df['lstat'] != df['lstat_iqr'], ['lstat', 'lstat_iqr']])
```
